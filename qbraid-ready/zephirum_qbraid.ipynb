```python
import zephirum, json

counts = {"00": 521, "11": 503}
par = sum(v for k, v in counts.items() if k.count("1") % 2 == 0)
t = sum(counts.values())
v = zephirum.decidir({"par": par, "impar": t - par}, shots=t, threshold=0.9, target_key="par")
print("COUNTS", json.dumps(counts, sort_keys=True))
print("VERDICT", v["verdict"], "p_hat", v["p_hat"], "epsilon", v["epsilon"])
print("RECEIPT_OK", zephirum.verificar_recibo(v))
print("RECEIPT", json.dumps(v, sort_keys=True))
```

```python
import zephirum
casos = [
    ("bell_paridade", {"par": 2024, "impar": 24}, 0.9, "par"),
    ("dj_constante", {"zero": 2048, "resto": 0}, 0.7, "zero"),
    ("dj_balanceada", {"zero": 0, "resto": 2048}, 0.7, "zero"),
]
for nome, contagens, limiar, alvo in casos:
    r = zephirum.decidir(contagens, shots=sum(contagens.values()), threshold=limiar, target_key=alvo)
    print(nome, r["verdict"], r["p_hat"], zephirum.verificar_recibo(r))
```
